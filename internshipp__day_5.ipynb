```python
import pandas as pd
import numpy as np

# -----------------------------
# Load Dataset
# -----------------------------
df = pd.read_csv("/content/archive (3).zip")

print("Original Shape:", df.shape)

# -----------------------------
# Remove Duplicate Rows
# -----------------------------
df = df.drop_duplicates()

# -----------------------------
# Remove Extra Spaces from Column Names
# -----------------------------
df.columns = df.columns.str.strip()

# -----------------------------
# Handle Missing Values
# -----------------------------
# Numerical columns
numeric_columns = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_columns:
    df[col] = df[col].fillna(df[col].mean())

# Categorical columns
categorical_columns = df.select_dtypes(include=['object']).columns

for col in categorical_columns:
    # Remove extra spaces
    df[col] = df[col].astype(str).str.strip()

    # Replace empty strings with NaN
    df[col] = df[col].replace('', np.nan)

    # Fill missing values with mode
    if df[col].mode().empty == False:
        df[col] = df[col].fillna(df[col].mode()[0])

# -----------------------------
# Remove Outliers using IQR
# -----------------------------
for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

# -----------------------------
# Reset Index
# -----------------------------
df.reset_index(drop=True, inplace=True)

# -----------------------------
# Check Missing Values
# -----------------------------
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nCleaned Dataset Shape:", df.shape)

# -----------------------------
# Save Cleaned Dataset
# -----------------------------
df.to_csv("cleaned_tourism_recommendation_dataset.csv", index=False)

print("\nDataset cleaned successfully!")
print("Saved as: cleaned_tourism_recommendation_dataset.csv")
```

```python
# Target variable
target = "rating"

# Features and target
X = df.drop(columns=[target])
y = df[target]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

# Encode categorical columns
le = LabelEncoder()

for col in X.select_dtypes(include='object').columns:
    X[col] = le.fit_transform(X[col].astype(str))

print(X.head())
```

```python
# Select features with correlation > 0.10
selected_features = corr_matrix[abs(corr_matrix) > 0.10].index

# Remove target column
selected_features = selected_features.drop(target)

print("Selected Features:")
print(selected_features)

# Create new feature dataset
X = X[selected_features]

print("New Feature Shape:", X.shape)
```

```python
from sklearn.feature_selection import SelectKBest, f_regression

# Select top 5 features
selector = SelectKBest(score_func=f_regression, k=5)

X_selected = selector.fit_transform(X, y)

selected_columns = X.columns[selector.get_support()]

print("Top 5 Selected Features:")
print(selected_columns)
```

```python
final_df = pd.DataFrame(X_selected, columns=selected_columns)
final_df['Rating'] = y.values

final_df.to_csv("selected_features_dataset.csv", index=False)

print("Feature-selected dataset saved successfully!")
```

#train test split

```python
from sklearn.model_selection import train_test_split

# -----------------------------
# Define Features and Target
# -----------------------------
X = encoded_df.drop("rating", axis=1)   # Features
y = encoded_df["rating"]                # Target

# -----------------------------
# Train-Test Split
# -----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,      # 20% testing data
    random_state=42,     # For reproducibility
    shuffle=True
)

# -----------------------------
# Display Shapes
# -----------------------------
print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)
print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)
```

```python
print("\nFirst 5 rows of Training Features:")
print(X_train.head())

print("\nFirst 5 rows of Testing Features:")
print(X_test.head())

print("\nFirst 5 Training Target Values:")
print(y_train.head())
```

#Model training

```python
from sklearn.ensemble import RandomForestRegressor

# -----------------------------
# Train the Model
# -----------------------------
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# Fit the model
model.fit(X_train, y_train)

print("Model trained successfully!")
```

```python
# -----------------------------
# Make Predictions
# -----------------------------
y_pred = model.predict(X_test)

print("First 10 Predictions:")
print(y_pred[:10])
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

# R² Score
r2 = r2_score(y_test, y_pred)

# MAE
mae = mean_absolute_error(y_test, y_pred)

# MSE
mse = mean_squared_error(y_test, y_pred)

# RMSE
rmse = np.sqrt(mse)

print("R² Score :", r2)
print("MAE      :", mae)
print("MSE      :", mse)
print("RMSE     :", rmse)
```
